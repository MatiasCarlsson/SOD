```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
metrics_dir = "../data/metrics"
figures_dir = "figures"
os.makedirs(figures_dir, exist_ok=True)
```

### Experimento E1: Rendimiento según Volumen de Datos (Python vs Hadoop vs Spark)

```python
df_e1 = pd.read_csv(os.path.join(metrics_dir, "experiment_1_volume.csv"))
agg_e1 = df_e1.groupby(["dataset", "engine"])["execution_time_sec"].agg(["mean", "std", "count"]).reset_index()
display(agg_e1)

plt.figure(figsize=(9, 5))
sns.barplot(data=df_e1, x="dataset", y="execution_time_sec", hue="engine", palette="Set2")
plt.title("Figura 1. Tiempo de Ejecución por Motor y Volumen (E1)", loc="left", fontweight="bold")
plt.ylabel("Tiempo (segundos)")
plt.xlabel("Conjunto de Datos")
plt.show()
```

### Experimento E2: Escalabilidad y Aceleración (Speedup) en Apache Spark

```python
df_e2 = pd.read_csv(os.path.join(metrics_dir, "experiment_2_spark_scalability.csv"))
agg_e2 = df_e2.groupby("workers")["compute_time_sec"].mean().reset_index()
t1 = agg_e2.loc[agg_e2["workers"] == 1, "compute_time_sec"].values[0]
agg_e2["speedup"] = t1 / agg_e2["compute_time_sec"]
agg_e2["efficiency"] = (agg_e2["speedup"] / agg_e2["workers"]) * 100
display(agg_e2)

plt.figure(figsize=(8, 4.5))
plt.plot(agg_e2["workers"], agg_e2["speedup"], marker="o", label="Speedup Real")
plt.plot(agg_e2["workers"], agg_e2["workers"], linestyle="--", color="gray", label="Speedup Ideal")
plt.title("Figura 2. Curva de Aceleración en Apache Spark (E2)", loc="left", fontweight="bold")
plt.xlabel("Workers (p)")
plt.ylabel("Speedup (Sp)")
plt.legend()
plt.show()
```

### Experimento E5: Latencia de Control en Lazo Cerrado

```python
df_e5 = pd.read_csv(os.path.join(metrics_dir, "experiment_5_latency_summary.csv"))
display(df_e5)

plt.figure(figsize=(8, 4.5))
plt.errorbar(df_e5["sensor_count"], df_e5["mean_latency_ms"], yerr=df_e5["std_latency_ms"], fmt="-o", label="Latencia Media ± std", color="crimson")
plt.plot(df_e5["sensor_count"], df_e5["p95_latency_ms"], "--s", label="P95", color="darkorange")
plt.title("Figura 3. Latencia del Controlador según Concurrencia de Sensores (E5)", loc="left", fontweight="bold")
plt.xlabel("Cantidad de Sensores Simulados")
plt.ylabel("Latencia (ms)")
plt.legend()
plt.show()
```
